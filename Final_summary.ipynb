# **Final Summary**

**Problem:** Predict telecom churn — customer leave or stay.

**Approach:** Clean data (fix TotalCharges, handle missing). Encode categorical cols (one-hot). Scale numeric (StandardScaler). Split train/test. Train Logistic Regression + Decision Tree. Bonus: K-Means segment customers.

**Best model:** Decision Tree — recall priority, catch more actual churners vs plain accuracy.

**Key business insight:** `tenure` biggest driver (~ 0.42 importance) — new customers churn most, long-tenure customers stay. `Fiber optic` internet close 2nd (~ 0.36) — fiber customers churn far more than DSL/no-internet, likely price or service quality issue. TotalCharges, Electronic check payment, MonthlyCharges also matter but much smaller weight. Two-year contract + TechSupport_Yes → lower churn risk (protective factors).

**Next step:** Target retention offers at low-tenure + fiber-optic customers first — highest churn risk segment. Investigate why fiber optic churn high (price? service issue?). Consider incentivize switch from electronic check → auto-pay methods, correlate w/ lower churn.
