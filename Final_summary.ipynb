# **Final Summary**

**Problem:** Predict telecom churn — customer leave or stay.

**Approach:** Clean data (fix TotalCharges, handle missing). Encode categorical cols (one-hot). Scale numeric (StandardScaler). Split train/test. Train Logistic Regression + Decision Tree. Bonus: K-Means segment customers.

**Best model:** Compare recall, not just accuracy — missed churner cost more than false alarm. Pick model w/ higher recall on churn class. State which (Logistic or Tree) + score.

**Key business insight:** Month-to-month contract customers churn far more than 1yr/2yr contracts. High MonthlyCharges + low tenure = high churn risk group.

**Next step:** Target retention offers at month-to-month, high-charge, low-tenure segment first — biggest ROI for reducing churn.
